```python
import tensorflow as tf
from tensorflow import keras

import matplotlib.pyplot as plt
import numpy as np
import sys
import pandas as pd
import cv2 as cv
from sklearn.model_selection import train_test_split
```

```python
print(tf.__version__)
print(keras.__version__)
```

```python
data = pd.read_json (r'images/otsu-images/dataframe.json')
print(data.columns)
```

```python
labels = data.iloc[:, 2:] # coordinates of corners
# y = np.ravel(labels) # converts into an array (but then it loses structure?...)
labels
```

```python
X = [] # contains all the image data
for img_path in data.imagePath:
    img = cv.imread(str(img_path))
    resized_image = cv.resize(img, (540, 720))

    imageArray = np.asarray(resized_image)
    print(img_path)
    X.append(imageArray)
```

```python
X = np.asarray(X)
# X 
X.shape
# images, height, width, rgb values of pixel
```

```python
X_train_full, X_test, y_train_full, y_test = train_test_split(X, labels, test_size=0.2, random_state=42)

print(X_train_full.shape)
print(X_test.shape)
```

```python
X_valid, X_train = X_train_full[:300], X_train_full[300:]
y_valid, y_train = y_train_full[:300], y_train_full[300:]

print(X_valid.shape)
# print(y_valid.shape) # 12 rows for bbox values
```

```python
keras.backend.clear_session()
np.random.seed(42)
tf.random.set_seed(42)
```

```python
number_of_bbox_values = labels.shape[1] # bbox means bounding box

# copied from classification model (needs to be changed)
model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[720, 540, 3]),
#     keras.layers.Dense(30, activation="relu", input_shape=(540, 720)),
    keras.layers.Dense(30, activation="relu"),
#     keras.layers.Dense(300, activation="relu"),
    keras.layers.Dense(100, activation="relu"),
#     keras.layers.Dense(number_of_bbox_values, activation="softmax")
    keras.layers.Dense(number_of_bbox_values, activation="sigmoid")
])
```

```python
model.summary()
```

```python
model.compile(loss="categorical_crossentropy",
              optimizer="sgd",
              metrics=["accuracy"])

# optimizers: "sgd" (Stochastic gradient descent), "Adagrad" (also gradient descent)
# loss: "sparse_categorical_crossentropy", "categorical_crossentropy"
```

```python
# print(np.ravel(X_valid).shape)
# print(y_valid.shape)
print(X_train.shape)
print(y_train.shape)

# history = model.fit(X_train, y_train, epochs=2, validation_data=(X_valid, y_valid))
# history = model.fit(X_train, y_train, epochs=2, validation_data=(np.ravel(X_valid), np.ravel(y_valid)))
history = model.fit(X_train, y_train, epochs=2, validation_data=(X_valid, y_valid))
```

```python
model.evaluate(X_test, y_test)
```
